Import Dataset and do a quick Inspection

```python
import urllib.request
import pandas as pd

# Download the dataset from UCI Repository
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00292/Wholesale%20customers%20data.csv"
urllib.request.urlretrieve(url, "wholesale_customers.csv")

# Load into a DataFrame
df = pd.read_csv("wholesale_customers.csv")

# Quick inspection
print("Shape:", df.shape)
print("\nFirst 5 rows:")
print(df.head())
print("\nColumn names:", df.columns.tolist())
print("\nData types:\n", df.dtypes)
print("\nBasic statistics:\n", df.describe())
print("\nMissing values:\n", df.isnull().sum())
```

Step 1: Data Understanding (10 points)


```python
import matplotlib.pyplot as plt
import seaborn as sns

spending_cols = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']

fig, axes = plt.subplots(2, 3, figsize=(15, 8))
axes = axes.flatten()

for i, col in enumerate(spending_cols):
    axes[i].hist(df[col], bins=30, color='steelblue', edgecolor='black')
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel('Annual Spending')
    axes[i].set_ylabel('Frequency')

plt.suptitle('Spending Distributions (Raw)', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```
