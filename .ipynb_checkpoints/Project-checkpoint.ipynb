Next we begin to normalize the data.

```python
Step 3: 
```

```python
import numpy as np

spending_cols = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']

# Apply log transformation on the spending columns
df_log = df[spending_cols].apply(np.log)

# Check for NaN and Infinite values after the transformation
print("NaN values in df_log:")
print(df_log.isna().sum())

print("\nInfinite values in df_log:")
print(np.isinf(df_log).sum())


# Plot before vs after for comparison
fig, axes = plt.subplots(2, 6, figsize=(20, 8))

for i, col in enumerate(spending_cols):
    # Before
    axes[0, i].hist(df[col], bins=30, color='salmon', edgecolor='black')
    axes[0, i].set_title(f'{col}\n(Raw)', fontsize=9)
    axes[0, i].set_xlabel('Spending')
    axes[0, i].set_ylabel('Frequency')

    # After
    axes[1, i].hist(df_log[col], bins=30, color='steelblue', edgecolor='black')
    axes[1, i].set_title(f'{col}\n(Log Transformed)', fontsize=9)
    axes[1, i].set_xlabel('Log Spending')
    axes[1, i].set_ylabel('Frequency')

plt.suptitle('Before vs After Log Transformation', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

Why log transformation?
The raw spending data is heavily right-skewed — a few extreme spenders stretch the scale massively. This causes two problems for K-Means:

Outliers dominate — cluster centers get pulled toward extreme values instead of representing the typical customer
Distance is distorted — K-Means uses Euclidean distance, so a customer spending $100k on Fresh looks impossibly far from everyone else, even if their overall pattern is similar

np.log() compresses that long right tail, making the data much more symmetric and bell-shaped. Note, because log(0) will yield undefined, we check for NaN and Infinite values again. An alternate method is using np.log1p() which is defined to safely handle zero values.

Next we have to scale the results to be within 0 and 1 for our model to be able to process the data.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(df_log)

print("Mean of each column (should be ~0):", X_scaled.mean(axis=0).round(2))
print("Std of each column (should be ~1):", X_scaled.std(axis=0).round(2))
```

**K-Means** uses **Euclidean distance** to assign points to clusters. If one feature has a wider numeric range than another, it will dominate the distance calculation — effectively "outvoting" the other features, even if they're equally important for finding customer segments.

The above results illustrate that average vale for each column is ~0 and the standard deviation is 1.

This puts all 6 spending categories on equal footing, so no single feature dominates clustering just because of its raw scale.

### Step 4 — the Elbow Method to choose the right number of clusters (K).

```python
from sklearn.cluster import KMeans

wcss = []
for k in range(1, 11):
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)

plt.figure(figsize=(8, 5))
plt.plot(range(1, 11), wcss, marker='o', linestyle='-', color='steelblue')
plt.title('Elbow Method for Optimal K', fontsize=13, fontweight='bold')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('WCSS (Within-Cluster Sum of Squares)')
plt.xticks(range(1, 11))
plt.grid(alpha=0.3)
plt.show()
```

Based on the graph above the graph above, there is no sharp turn at any point, but after K = 2 or 3, the returns diminish significantly. Initial thesis is K = 3. Next we will use the Silhouette Score to validate our hypothesis.

### Step 5: Hierarchical Clustering (10 pts)

```python
from scipy.cluster.hierarchy import linkage, dendrogram

linked = linkage(X_scaled, method="ward")

plt.figure(figsize=(10,5))
plt.title('Dendrogram for Optimal K', fontsize=13, fontweight='bold')
dendrogram(linked)
plt.show()
```

### Step 6 Cluster Evaluation (10 pts)

```python
from sklearn.metrics import silhouette_score

sil_scores = []
k_range = range(2, 11)  # silhouette score requires at least 2 clusters

for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    labels = kmeans.fit_predict(X_scaled)
    score = silhouette_score(X_scaled, labels)
    sil_scores.append(score)
    print(f"K={k}: Silhouette Score = {score:.4f}")

plt.figure(figsize=(8, 5))
plt.plot(list(k_range), sil_scores, marker='o', linestyle='-', color='coral')
plt.title('Silhouette Score by Number of Clusters', fontsize=13, fontweight='bold')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Silhouette Score')
plt.xticks(list(k_range))
plt.grid(alpha=0.3)
plt.show()
```

The Silhouette score measures how well separated each cluster is. The best scores are at K=2 and K=3 as we predicted in the Elbow Method above. The Dendrogram solidifies our choice for K = 2.

### Step 4 — Final K-Means Model with K=2

```python
kmeans = KMeans(n_clusters=2, random_state=42)
labels = kmeans.fit_predict(X_scaled)

# Add cluster labels back to the original dataframe
df['Cluster'] = labels

print(df['Cluster'].value_counts())
```

```python
spending_cols = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']

for col in spending_cols:
    # Apply pd.cut to the log-transformed column instead of the raw column
    df[f'{col}Group'] = pd.cut(df_log[col], bins=3, labels=["Low", "Medium", "High"])

# Check the result
print(df[[f'{col}Group' for col in spending_cols]].head())

# Check how many customers fall into each bin per category
for col in spending_cols:
    print(f"\n{col}Group distribution:")
    print(df[f'{col}Group'].value_counts())
```

```python
spending_cols = ['Fresh', 'Milk', 'Grocery', 'Frozen', 'Detergents_Paper', 'Delicassen']

for col in spending_cols:
    df[f'{col}Group'] = pd.qcut(df_log[col], q=3, labels=["Low", "Medium", "High"])

# Check distribution
for col in spending_cols:
    print(f"\n{col}Group distribution:")
    print(df[f'{col}Group'].value_counts())
```

This last method more evenly distributes the customers. **qcut** divides the group by percentile rank versus raw values thus giving us an even distribution.

### Step 8: Interpretation & Insights (10 pts)

In addition to the insights already declared, let's look for some associations between customers and what is purchased, like we saw in the heatmap. To do this we will use a new tool to find these associations.

```python
# Build the Transaction Dataset
from mlxtend.frequent_patterns import apriori, association_rules

# Combine all group columns into one transaction dataframe
group_cols = [f'{col}Group' for col in spending_cols]

# One-hot encode the categorical groups
df_transactions = pd.get_dummies(df[group_cols])

print(df_transactions.shape)
print(df_transactions.head())
```

What has been done above is to reformat the information we have on the customers and the categories we broke them into and reformat it into a truth table. Each original spending column is now represented by three columns, for example, fresh_low, frsh_medium, fresh_high. Then based on the step we did before they would get true for fresh_low if that customer was categorized as low, while the remaining two columns would be zero. This format is what **apriori** uses as input when we start searching for associations between product purchases.

Below we set the parameters to use when starting the search:
* **min_support=0.3** — only keep itemsets that appear in at least 30% of customers (132+ customers). This filters out rare, unreliable combinations
* **metric="lift"** — rank rules by lift, since we care about genuine relationships, not just common co-occurrences
* **min_threshold=1.2** — only keep rules where lift > 1.2, meaning the relationship is at least 20% stronger than random chance

```python
# Find frequent itemsets
frequent_itemsets = apriori(df_transactions, min_support=0.3, use_colnames=True)

# Generate association rules
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.2)

# Sort by lift descending
rules = rules.sort_values("lift", ascending=False)

rules['antecedents'] = rules['antecedents'].apply(lambda x: ', '.join(list(x)))
rules['consequents'] = rules['consequents'].apply(lambda x: ', '.join(list(x)))

print(f"Number of frequent itemsets: {len(frequent_itemsets)}")
print(f"Number of rules found: {len(rules)}")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].to_string(index=False))
```

And we got nothing. Looking at the frequency table of the groups we defined earlier.

```python
print(f"Number of frequent itemsets: {len(frequent_itemsets)}")
print(frequent_itemsets.sort_values('support', ascending=False).head(20))
```

With 18 binary columns (6 categories × 3 levels) and balanced bins (~33% each), the probability of any specific combination of items appearing together drops quickly. For example:
P(Fresh_High AND Grocery_High) ≈ 0.33 × 0.33 = ~0.11

So we will lower the level and try again.

```python
# Re-run with min_support=0.1
frequent_itemsets = apriori(df_transactions, min_support=0.1, use_colnames=True)
rules = association_rules(frequent_itemsets, metric="lift", min_threshold=1.2)
rules = rules.sort_values("lift", ascending=False)

# Clean up display
rules['antecedents'] = rules['antecedents'].apply(lambda x: ', '.join(list(x)))
rules['consequents'] = rules['consequents'].apply(lambda x: ', '.join(list(x)))

print(f"Number of frequent itemsets: {len(frequent_itemsets)}")
print(f"Number of rules found: {len(rules)}")
print(rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']].to_string(index=False))
```

Association rule is always written as:
IF [antecedent] THEN [consequent]

Antecedent = the "IF" part — the condition you already know about a customer
Consequent = the "THEN" part — what you can predict about that customer

So in the above rules that were mined we see, for example:
If they have high spending on Delicassen and high spending on Detergents_Paper then there is a high proability that they will also have high spending on Milk and Grocery as well. 3 times more likely to purchase than random selection.

**Retail Bundle (Lift ~2.5–3.1):** The strongest rules consistently involved
high spending across Grocery, Milk, Detergents_Paper, and Delicassen together.
The bidirectional rule between Grocery_High and Detergents_Paper_High —
present in 28% of all customers with 84% confidence and lift of 2.50 —
represents the most widespread and reliable pattern in the dataset. These
rules collectively characterize a retail purchasing profile where shelf-stable
goods and household products are purchased as a consistent bundle.

**Retail Signal via Fresh Absence (Lift ~2.94):** The rule
`Fresh_Low + Grocery_High → Milk_High + Detergents_Paper_High` (confidence
71%) reveals that low Fresh spending combined with high Grocery spending is
a strong predictor of retail behavior. The absence of Fresh purchasing is
itself a meaningful signal — retail stores do not purchase perishables for
cooking.

**Horeca Signal (Lift ~1.47):** The bidirectional rule between Fresh_High and
Frozen_High, while weaker than the retail patterns, reflects the food-service
purchasing profile where both fresh ingredients and frozen goods are needed
for meal preparation. The weaker lift (1.47 vs 3.0+) is consistent with
greater spending diversity among Horeca customers.

So what does this all mean? Promotions on Frozen Items should be targeted to Horeca customers and can be paired with marketing material where they purchase vegetables and fruits and vice versa. Sales on fresh items may not be a strong pull for Retail customers and money can be better spent promoting grocery items to Retail.



```python
import matplotlib.pyplot as plt
import matplotlib.cm as cm
import numpy as np

plt.figure(figsize=(10, 6))

scatter = plt.scatter(
    rules['support'],
    rules['confidence'],
    c=rules['lift'],
    cmap='RdYlGn',
    s=100,
    alpha=0.7,
    edgecolors='black',
    linewidths=0.5
)

plt.colorbar(scatter, label='Lift')
plt.title('Association Rules — Support vs Confidence (colored by Lift)',
          fontsize=13, fontweight='bold')
plt.xlabel('Support (how common is this rule?)')
plt.ylabel('Confidence (how reliable is this rule?)')
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
top10 = rules.head(10).copy()

# Shorten rule labels for readability
top10['rule'] = top10['antecedents'] + ' → ' + top10['consequents']
top10['rule'] = top10['rule'].str.replace('Group_', ' ', regex=False)

plt.figure(figsize=(12, 7))
bars = plt.barh(range(len(top10)), top10['lift'],
                color=plt.cm.RdYlGn(top10['lift'] / top10['lift'].max()),
                edgecolor='black')

plt.yticks(range(len(top10)), top10['rule'], fontsize=8)
plt.xlabel('Lift', fontsize=11)
plt.title('Top 10 Association Rules by Lift', fontsize=13, fontweight='bold')
plt.axvline(x=1.0, color='red', linestyle='--', linewidth=1, label='Lift = 1 (random chance)')
plt.legend()
plt.gca().invert_yaxis()  # highest lift at top
plt.tight_layout()
plt.show()
```

## 5. Summary

Both unsupervised methods — K-Means clustering and association rule mining —
independently converged on the same underlying structure in the data: a
retail-oriented segment characterized by high spending on Grocery, Milk,
Detergents_Paper, and Delicassen, and a food-service-oriented segment
characterized by high spending on Fresh and Frozen products. The convergence
of two independent methods on the same segmentation, further validated by
near-perfect alignment with the known Channel variable, provides strong
evidence that these patterns reflect genuine customer behavioral differences
rather than artifacts of the analytical approach.
